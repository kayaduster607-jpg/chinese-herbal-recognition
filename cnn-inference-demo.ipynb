使用公开的 CNN 基线权重和数据集样本，实际执行一次前向推理。模型输出类别索引；原始类别映射未确认。

```python
import os
from pathlib import Path
from PIL import Image
from IPython.display import display, HTML
from infer_cnn import predict

# 环境变量只用于指定权重路径；不会读取项目环境文件。
model_path = Path(os.environ.get("HERBAL_MODEL_PATH", "weights/chinese_medicine_classifier_simple.h5"))
image_path = Path("docs/images/sample-herb.jpg")
result = predict(model_path, image_path)

print("实际执行完成 | TensorFlow", result["tensorflow"])
print("模型输入:", result["input_shape"], "| 输出维度:", result["output_dimensions"])
print("预处理：RGB → 150 × 150 → 0–1 缩放（原 CNN 预处理待核实）")
print("类别索引尚未核实，以下仅展示输出索引与模型分数：")
for item in result["top5"]:
    print(f"  索引 {item['index']:>2}   分数 {item['score']:.4f}")

print("\n本次输入的真实数据集图片：")
display(Image.open(image_path).resize((300, 300)))
```
